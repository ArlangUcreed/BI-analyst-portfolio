### 2.2. Типы данных

- Если встречаются некорректные типы данных, предположите их причины.
- При необходимости проведите преобразование типов данных. Помните, что столбцы с числовыми данными и пропусками нельзя преобразовать к типу `int64`. Сначала вам понадобится обработать пропуски, а затем преобразовать типы данных.

```python
# Определяем числовые столбцы с правильными именами
numeric_columns = ['year_of_release', 'na_sales', 'eu_sales', 'jp_sales', 'other_sales', 'critic_score', 'user_score']

# Применяем обработку и преобразование типов для всех числовых столбцов
for column in numeric_columns:
    # Заменяем 'unknown' и 'N/A' на NaN
    data[column] = data[column].replace({'unknown': pd.NA, 'N/A': pd.NA})  
    
    # Проверка на наличие некорректных типов данных
    if data[column].dtype == 'object':
        # Используем display для отображения уникальных значений
        display(data[column].unique())  # Отображаем уникальные значения напрямую
        
    # Преобразуем в числовой тип, заменяя некорректные значения на NaN
    data[column] = pd.to_numeric(data[column], errors='coerce')  # Преобразуем в числовой тип

# Обработка пропусков
for column in numeric_columns:
    # Заменяем NaN на 0
    data[column].fillna(0, inplace=True)

# Проверка результата
display(data[numeric_columns].info())
```

- В числовых столбцах могут встретиться строковые значения, например `unknown` или другие. Приводите такие столбцы к числовому типу данных, заменив строковые значения на пропуски.

```python
# Обработка строковых значений и преобразование
for column in numeric_columns:
    # Проверка на наличие строковых значений
    if data[column].dtype == 'object':
        display(f"Уникальные значения в столбце '{column}': {data[column].unique()}")

    # Преобразуем в числовой тип, заменяя некорректные значения на NaN
    data[column] = pd.to_numeric(data[column], errors='coerce')

# Проверка результата
display(data[numeric_columns])
```

# Для нормализации текстовых данных в данном наборе данных, используются методы для приведения строк к единому регистру и удаления лишних пробелов. Это помогает устранить неявные дубликаты, связанные с опечатками или разным способом написания.

# Шаги для нормализации данных:

- Приводятся названия жанров, платформ и рейтингов к нижнему регистру. 
- Удаляются лишние пробелы в начале и конце строк.
- Проверяюся уникальные значения после нормализации.

    Объяснение кода:

- str.lower() и str.upper() приводят строки к нижнему и верхнему регистру соответственно.
- str.strip() удаляет пробелы в начале и конце строк.
- unique() выводит уникальные значения в каждом из столбцов для проверки.
- nunique() показывает количество уникальных значений до и после нормализации, что помогает выявить неявные дубликаты.

```python
# Выводим список всех столбцов в DataFrame для диагностики
print("Список столбцов в DataFrame:", data.columns.tolist())

# Нормализация текстовых данных
if 'Genre' in data.columns:
    data['Genre'] = data['Genre'].str.lower().str.strip()
else:
    print("Столбец 'Genre' не найден.")

if 'Platform' in data.columns:
    data['Platform'] = data['Platform'].str.lower().str.strip()
else:
    print("Столбец 'Platform' не найден.")

if 'Rating' in data.columns:
    data['Rating'] = data['Rating'].str.upper().str.strip()
else:
    print("Столбец 'Rating' не найден.")

# Проверка на явные дубликаты
duplicates = data.duplicated()
num_duplicates = duplicates.sum()

print(f"Количество явных дубликатов в данных: {num_duplicates}")

# Если дубликаты найдены, то можно их удалить
if num_duplicates > 0:
    data = data[~duplicates]  # Удаление дубликатов
    print("Явные дубликаты удалены.")
else:
    print("Явных дубликатов не найдено.")

# Проверяем количество строк после удаления дубликатов
print(f"Количество строк в данных после удаления дубликатов: {len(data)}")

# Для более удобного восприятия можно отобразить первые несколько строк DataFrame
display(data.head())
```
